#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
# Mean, min, and max of every numeric column with the missing values dropped
summary_stats <- data |>
    summarise(across(where(is.numeric),
        list(mean = ~ mean(.x, na.rm = TRUE),
            min = ~ min(.x, na.rm = TRUE),
            max = ~ max(.x, na.rm = TRUE)),
        .names = "{.col}_{.fn}")) |>
# Make into one tidy table with one row per variable
    pivot_longer(everything(),
        names_to = c("variable", "statistic"),
        names_pattern = "(.*)_(mean|min|max)") |>
    pivot_wider(names_from = statistic, values_from = value)
# Show table
summary_stats
```

<div style="background-color: #f0f0f0; padding: 10px;">
    
#### Task 4.2: Determine the amount of missingness (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates the number of missing values for each column in the data set (both numeric and categorical).
- Show the outputted table. 
</div>

```r
# Count the missing values in every column
missing_values <- data |>
    summarise(across(everything(), ~ sum(is.na(.x)))) |>
# Make table easy to read with one row per column
    pivot_longer(everything(), names_to = "column", values_to = "n_missing")
# Show table
missing_values
```
